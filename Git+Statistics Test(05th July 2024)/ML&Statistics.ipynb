```python
from google.colab import files
import pandas as pd
import matplotlib.pyplot as plt
import os
import seaborn as sns
import numpy as np

if not os.path.exists('sample_dataset1.xlsx'):
  uploaded = files.upload()
  for filename in uploaded.keys():
    with open(filename, 'wb') as f:
      f.write(uploaded[filename])
  print(f"Uploaded {list(uploaded.keys())[0]} and saved as sample_dataset1.xlsx")
else:
  print("sample_dataset1.xlsx already exists")

df = pd.read_excel('sample_dataset1.xlsx')
print(df)
print(df.isna())

df = df.interpolate(method='linear', axis=0)
print(df)

plt.figure(figsize=(25, 15))
sns.barplot(x='County', y='Total Loan Amount', data=df)
plt.title('Total Loan Amount by each County')
plt.xlabel('County')
plt.ylabel('Amount')
plt.show()

plt.figure(figsize=(25, 15))
plt.plot(df['County'], df['Total Loan Amount'], label='Total Loan Amount', marker='o')
plt.plot(df['County'], df['DPA Loan Amount'], label='DPA Loan Amount', marker='s')
plt.legend()
plt.title('Total Loan Amount and DPA Loan Amount by each County')
plt.xlabel('County')
plt.ylabel('Amount')
plt.show()

plt.figure(figsize=(25, 25))
pie_wedges, texts, autotexts = plt.pie(df['Mortgages Financed'], autopct='%1.2f%%', startangle=90)
plt.legend(pie_wedges, df['County'], loc='best')
plt.title('Mortgages Financed by each County')
plt.axis('equal')
plt.show()

data = df['No. of People staying in the county']
bin_edges = np.arange(0, max(data)+1)
plt.hist(data, bins=bin_edges, edgecolor='black', align='mid', histtype='bar', rwidth=0.8)
plt.title('No. of People staying in counties')
plt.xlabel('No. of People')
plt.ylabel('Frequency')
plt.show()
```

Use of

*   Pandas:- To create and manipulate dataframes
*   numpy:- To create and manipulate arrays and matrices, and for scientific purposes


*   seaborn :- For drawing attractive, informative and complex statistical plots, in fewer lines of code than matplotlib. Built on top of matplotlib.
*   Matplotlib:- To create various kinds of static and animated plots like bar chart, line chart, pie chart, histogram, etc.


*   OS :- Provides a way for utilising operating system - dependent functionality, like reading and writing into files.
